```python
import pandas as pd

def sort_and_save_data(input_filename, output_filename):
    """
    Reads the BMW sales data, sorts the rows by Model and then by Year, 
    and saves the result to a new CSV file.
    
    Args:
        input_filename (str): The name of the input CSV file.
        output_filename (str): The name of the output CSV file.
    """
    try:
        # 1. Read the CSV file into a DataFrame
        df = pd.read_csv(input_filename)

        # 2. Sort the DataFrame
        # Sort first by 'Model' (alphabetic) and then by 'Year' (chronological)
        # This groups all rows of the same model together, ordered by year.
        df_sorted = df.sort_values(by=['Model', 'Year'], ascending=[True, True])

        # 3. Save the sorted DataFrame to a new CSV file
        df_sorted.to_csv(output_filename, index=False)
        
        print(f"Successfully sorted data and saved to {output_filename}")
        print(f"New file contains {len(df_sorted)} rows.")

    except FileNotFoundError:
        print(f"Error: The input file '{input_filename}' was not found.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Define the file names
input_file = 'BMW sales data (2010-2024) (1).csv'
output_file = 'bmw_sorted_by_model_year.csv'

# Execute the function
sort_and_save_data(input_file, output_file)
```

```python
import pandas as pd

def consolidate_and_save_data(input_filename, output_filename):
    """
    Reads the BMW sales data, groups rows based on all primary categorical 
    and configuration columns, calculates the mean of numerical columns 
    within each group, and saves the consolidated result.
    
    Args:
        input_filename (str): The name of the input CSV file.
        output_filename (str): The name of the output CSV file.
    """
    try:
        # 1. Read the CSV file into a DataFrame
        df = pd.read_csv(input_filename)

        # 2. Define the columns for grouping (The vehicle configuration keys)
        grouping_keys = [
            'Model', 
            'Year', 
            'Region', 
            'Color', 
            'Fuel_Type', 
            'Transmission', 
            'Engine_Size_L'
        ]

        # 3. Define the columns to aggregate (The numerical metrics)
        aggregation_cols = ['Mileage_KM', 'Price_USD', 'Sales_Volume']

        # 4. Group the data and calculate the mean for the numerical columns
        # The result is a new DataFrame where the grouping keys become the index.
        df_consolidated = df.groupby(grouping_keys)[aggregation_cols].mean().reset_index()

        # 5. Round the resulting numerical columns for cleaner presentation
        # We round Mileage_KM and Price_USD to the nearest integer
        df_consolidated['Mileage_KM'] = df_consolidated['Mileage_KM'].round(0).astype(int)
        df_consolidated['Price_USD'] = df_consolidated['Price_USD'].round(0).astype(int)
        # Sales_Volume is kept as float/int based on how it was averaged
        df_consolidated['Sales_Volume'] = df_consolidated['Sales_Volume'].round(0).astype(int)

        # 6. Save the consolidated DataFrame to a new CSV file
        df_consolidated.to_csv(output_filename, index=False)
        
        print(f"Successfully consolidated data and saved to {output_filename}")
        print(f"Original rows: {len(df)}")
        print(f"Consolidated configurations: {len(df_consolidated)}")

    except FileNotFoundError:
        print(f"Error: The input file '{input_filename}' was not found.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Define the file names
input_file = 'BMW sales data (2010-2024) (1).csv'
output_file = 'bmw_consolidated_data.csv'

# Execute the function
consolidate_and_save_data(input_file, output_file)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

def create_sales_volume_graph(input_filename):
    """
    Filters the data for the years 2019-2024, aggregates total sales volume by year and region,
    and generates a multi-line graph with a line for each region using Matplotlib.
    """
    try:
        # 1. Load the data
        df = pd.read_csv(input_filename)
        
        # 2. Filter for the target years (2010 to 2024)
        df_filtered = df[(df['Year'] >= 2010) & (df['Year'] <= 2024)].copy()
        
        # 3. Aggregate: Calculate the total Sales_Volume for each Year AND Region
        df_sales_trend = df_filtered.groupby(['Year', 'Region'])['Sales_Volume'].sum().reset_index()
        
        # 4. Data Preparation: Convert Year to string for cleaner axis labels
        df_sales_trend['Year'] = df_sales_trend['Year'].astype(str)

        print("Data prepared for plotting (Sales Volume by Year and Region):")
        print(df_sales_trend.head(10).to_string(index=False))

        # 5. Create the Plot
        plt.figure(figsize=(12, 7))
        
        # Get unique regions to loop through and plot
        regions = df_sales_trend['Region'].unique()
        
        # Plot a line for each region
        for region in regions:
            # Filter data for the current region
            region_data = df_sales_trend[df_sales_trend['Region'] == region]
            
            # Plot the line
            plt.plot(
                region_data['Year'], 
                region_data['Sales_Volume'], 
                marker='o', 
                linestyle='-', 
                linewidth=2,
                label=region # Add the region name to the legend
            )
        
        # Add titles and labels
        plt.title('BMW Sales Volume Trend by Region (2019 - 2024)', fontsize=16)
        plt.xlabel('Year', fontsize=12)
        plt.ylabel('Total Sales Volume', fontsize=12)
        
        # Enhance readability
        plt.grid(axis='y', linestyle='--', alpha=0.7)
        plt.xticks(rotation=45)
        
        # Add the legend to distinguish the lines
        plt.legend(title='Region', loc='upper left')
        
        plt.tight_layout()
        plt.show()

    except FileNotFoundError:
        print(f"Error: The input file '{input_filename}' was not found.")
    except Exception as e:
        print(f"An unexpected error occurred during plotting: {e}")

# --- Execution Block ---
input_file = 'BMW sales data (2010-2024) (1).csv'
create_sales_volume_graph(input_file)
```
