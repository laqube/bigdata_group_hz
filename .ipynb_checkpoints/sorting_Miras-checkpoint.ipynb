```python
import pandas as pd

def sort_and_save_data(input_filename, output_filename):
    """
    Reads the BMW sales data, sorts the rows by Model and then by Year, 
    and saves the result to a new CSV file.
    
    Args:
        input_filename (str): The name of the input CSV file.
        output_filename (str): The name of the output CSV file.
    """
    try:
        # 1. Read the CSV file into a DataFrame
        df = pd.read_csv(input_filename)

        # 2. Sort the DataFrame
        # Sort first by 'Model' (alphabetic) and then by 'Year' (chronological)
        # This groups all rows of the same model together, ordered by year.
        df_sorted = df.sort_values(by=['Model', 'Year'], ascending=[True, True])

        # 3. Save the sorted DataFrame to a new CSV file
        df_sorted.to_csv(output_filename, index=False)
        
        print(f"Successfully sorted data and saved to {output_filename}")
        print(f"New file contains {len(df_sorted)} rows.")

    except FileNotFoundError:
        print(f"Error: The input file '{input_filename}' was not found.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

# Define the file names
input_file = 'BMW sales data (2010-2024) (1).csv'
output_file = 'bmw_sorted_by_model_year.csv'

# Execute the function
sort_and_save_data(input_file, output_file)
```
